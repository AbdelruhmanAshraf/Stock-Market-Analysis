```python

# Add S&P 500 as a benchmark
benchmark_ticker = '^GSPC'
benchmark_df = yf.download(benchmark_ticker, 
                           start=start_date, 
                           end=end_date, 
                           progress=False)['Close']
benchmark_df.name = 'Close_GSPC'
print(benchmark_df.head())
```

```python

# Merge the benchmark data with the existing comparison DataFrame
comp_full = pd.DataFrame({
    'AAPL': df['Close_AAPL'], 
    'MSFT': df['Close_MSFT'],
    'GOOG': df['Close_GOOG'],
    'AMZN': df['Close_AMZN'],
    'META': df['Close_META']
}).dropna() # Get the 5 stocks closing prices

# Add S&P 500 benchmark
comp_full['GSPC'] = benchmark_df

# Drop any rows where any stock or the benchmark is missing data
comp_full = comp_full.dropna()

print('*** Combined Data (Stocks + Benchmark) ***')
print(comp_full.tail())
```

الصيغة التي استخدمتها: $\text{Annualized Return} = (\text{Total Return Factor})^{\frac{252}{\text{n\_days}}} - 1$

```python
n_days_full = len(comp_full)

# Calculate Daily Returns
returns_full = comp_full.pct_change().dropna()

# 1. Annualized Return (CAGR)
annualized_return_full = (comp_full.iloc[-1] / comp_full.iloc[0]) ** (252 / n_days_full) - 1
print('*** Annualized Return (CAGR) ***')
print(annualized_return_full.apply(lambda x: f"{x:.2%}"))

```

```python
# 2. Annualized Volatility
annualized_vol_full = returns_full.std() * (252 ** 0.5)
print('\n*** Annualized Volatility (Risk) ***')
print(annualized_vol_full.apply(lambda x: f"{x:.2%}"))
```

الوزن (Weight) هو أهم عامل في تحليل المحفظة، وهو يمثل النسبة المئوية من إجمالي رأس المال المخصص لكل سهم داخل المحفظة.

الفائدة الجوهرية للوزن هي أنه يتحكم في العائد والمخاطر الإجمالية للمحفظة.

```python
# Define the weights for an equally weighted portfolio (5 assets)
weights = [0.20, 0.20, 0.20, 0.20, 0.20]  

# Confirm the sum of weights is 1 (100%)
print(f"Portfolio Weights: {weights}")
print(f"Sum of Weights: {sum(weights)}")
```

حساب العوائد اليومية للمحفظةسنقوم بحساب العائد اليومي للمحفظة باستخدام القانون:$$\text{Portfolio\_Returns} = \sum (\text{Stock\_Returns} \times \text{Weight})$$

```python
# Calculate the daily returns for the 5 stocks (excluding the benchmark GSPC)
portfolio_returns_df = comp_full[['AAPL', 'MSFT', 'GOOG', 'AMZN', 'META']].pct_change().dropna()

# Calculate the daily portfolio return by multiplying stock returns by their weights and summing them up
portfolio_returns_series = (portfolio_returns_df * weights).sum(axis=1)
portfolio_returns_series.name = 'Portfolio_Returns'

print('*** Daily Portfolio Returns (Head) ***')
print(portfolio_returns_series.head())
print('\n*** Daily Portfolio Returns (Tail) ***')
print(portfolio_returns_series.tail())
```

$$\text{الأداء التراكمي} = (1 + R_1) \times (1 + R_2) \times (1 + R_3) \times \dots \times (1 + R_n)$$

```python
n_days_port = len(portfolio_returns_series)

# 1. Annualized Portfolio Return (CAGR)
# To do this, we need the cumulative performance.
# We assume the portfolio starts at 1, and compound the daily returns.
portfolio_cumulative_performance = (1 + portfolio_returns_series).prod()

# Formula: (Cumulative Performance) ^ (252 / n_days) - 1
annualized_portfolio_return = (portfolio_cumulative_performance) ** (252 / n_days_port) - 1

# 2. Annualized Portfolio Volatility
annualized_portfolio_vol = portfolio_returns_series.std() * (252 ** 0.5)

print('*** Portfolio Annual Performance (Equally Weighted) ***')
print(f"Annualized Return (CAGR): {annualized_portfolio_return:.2%}")
print(f"Annualized Volatility: {annualized_portfolio_vol:.2%}")
```

```python

# Calculate cumulative performance starting at 100
portfolio_normalized_performance = (1 + portfolio_returns_series).cumprod() * 100

# Combine the portfolio performance with the GSPC normalized performance
comparison_df = pd.DataFrame({
    'Equally_Weighted_Portfolio': portfolio_normalized_performance,
    'S&P_500 (GSPC)': norm_full['GSPC'].iloc[1:] # Use the normalized benchmark data
}).dropna()

# Plot the comparison
fig_portfolio = px.line(comparison_df, 
                        title='Portfolio Performance vs S&P 500 Benchmark',
                        labels={'value': 'Normalized Price (Index 100)', 'Date': 'Date'})

fig_portfolio.update_xaxes(rangeslider_visible=True)
fig_portfolio.show()
```
